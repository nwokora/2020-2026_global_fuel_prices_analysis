### **Q4: Low-Income Countries' Affordability Analysis for Different Subsidy Categories.**

```python
df['afford_index'] = df['petrol_usd_liter'] / df['income_num']

subsidy_effect = df[df['income_num']==1].groupby('subsidy_num')['afford_index'].mean()
print("Low income affordability by subsidy level:\n", subsidy_effect.sort_index())
```

```python
low_income_countries = df[df['income_num'] == 1][['country', 'region', 'subsidy_num', 'afford_index']]

for subsidy in [1, 2, 3, 4]:
    top5 = (low_income_countries[low_income_countries['subsidy_num'] == subsidy]
        .groupby('country', as_index=False)['afford_index']
        .mean()
        .sort_values('afford_index', ascending=False)
        .head(5)
        .reset_index(drop=True))
    
    print(f"\nLow-income countries affordability by subsidy_num={subsidy}:")
    print(top5)
    
    top5.to_csv(f'top5_low_income_subsidy_{subsidy}.csv', index=False)
```

### **Q7: Average Fuel Prices by Year**

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')

df['avg_fuel_price'] = df[['petrol_usd_liter', 'diesel_usd_liter', 'lpg_usd_liter']].mean(axis=1)

yearly_fuel_summary = (df.groupby(df['date'].dt.year)
      .agg(avg_petrol_price=('petrol_usd_liter', 'mean'),
          avg_diesel_price=('diesel_usd_liter', 'mean'),
          avg_lpg_price=('lpg_usd_liter', 'mean'),
          avg_fuel_price=('avg_fuel_price', 'mean'))
      .reset_index()
      .rename(columns={'date': 'year'})
      .sort_values('year'))

print(yearly_fuel_summary)
```

**Year and average fuel price**

```python
yearly_fuel_summary = (df.groupby(df['date'].dt.year)
      .agg(avg_fuel_price=('avg_fuel_price', 'mean'))
        
      .reset_index()
      .rename(columns={'date': 'year'})
      .sort_values('year'))

print(yearly_fuel_summary)
```

```python
yearly_fuel_summary[['year', 'avg_fuel_price']].to_csv('yearly_fuel.csv', index=False)
```

### **Q8: Regional Income Level Analysis**

```python
df['income_level'] = df['income_level'].astype(str).str.strip()

counts = df.groupby(['region', 'income_level']).size().reset_index(name='count')

region_totals = counts.groupby('region')['count'].sum().reset_index(name='region_total')

counts = counts.merge(region_totals, on='region')
counts['share_pct'] = (counts['count'] / counts['region_total']) * 100

order = ['Low','Middle','High']
spread = counts.pivot(index='region', columns='income_level', values='share_pct').reindex(columns=order).fillna(0)

region_summary = region_totals.merge(spread.reset_index(), on='region')

region_summary = region_summary.sort_values(by='region', ascending=True).reset_index(drop=True)

print(region_summary)
```

```python
region_summary.drop(columns=['region_total']).to_csv('region_income_spread.csv', index=False)
```

### **Q9: Regional Subsidy Level Analysis**

```python
df['subsidy_level'] = df['subsidy_level'].astype(str).str.strip()

counts = df.groupby(['region','subsidy_level']).size().reset_index(name='count')

region_totals = counts.groupby('region')['count'].sum().reset_index(name='region_total')

counts = counts.merge(region_totals, on='region')
counts['share_pct'] = (counts['count'] / counts['region_total']) * 100

order = ['Low','Medium','High','Very High']
spread = counts.pivot(index='region', columns='subsidy_level', values='share_pct').reindex(columns=order).fillna(0)

region_summary = region_totals.merge(spread.reset_index(), on='region')

sort_col = 'Very High'
region_summary = region_summary.sort_values(by=sort_col, ascending=False).reset_index(drop=True)

print(region_summary)
```

```python
region_summary.drop(columns=['region_total']).to_csv('region_subsidy_spread.csv', index=False)
```
